# Huggingface Pipelines

```python
import numpy as np

from PIL import Image as PImage

from torch import cuda, float16 as tf16
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from transformers import pipeline

from utils.utils import show_object_predictions

DEVICE = "cuda" if cuda.is_available() else "cpu"
```

### Text Classifier

```python
classifier = pipeline(task="text-classification",
                      model="siebert/sentiment-roberta-large-english")
```

```python
classifier("this is the best major studio ever")
```

### Object Detection

```python
DETR_MODEL = "facebook/detr-resnet-50"
OWL_MODEL = "google/owlv2-base-patch16"

detr = pipeline(task="object-detection",
                model=DETR_MODEL,
                device=DEVICE)

owl = pipeline(task="zero-shot-object-detection",
               model=OWL_MODEL,
               device=DEVICE)
```

```python
img = PImage.open("../imgs/desk.jpg")
img.thumbnail((500, 500))

res = detr(img)
show_object_predictions(img, res)
```

```python
img = PImage.open("../imgs/desk.jpg")
img.thumbnail((500, 500))

res = owl(img, ["poster", "pen"], threshold=0.12)
show_object_predictions(img, res)
```

### Image Processing

```python
DPT_MODEL = "Intel/dpt-large"

dpt = pipeline(task="depth-estimation",
               model=DPT_MODEL,
               device=DEVICE)
```

```python
img = PImage.open("../imgs/beatles.jpg")
img.thumbnail((500, 500))

res = dpt(img)
display(res["depth"])
```

```python
mask = np.array(res["depth"])
img_np = np.array(img)

ppix = (mask > 150) & (mask < 220)
img_np[~ppix] = [0,0,0]
PImage.fromarray(img_np)
```

### Translation

```python
PTEN_MODEL = "unicamp-dl/translation-pt-en-t5"

tokenizer = AutoTokenizer.from_pretrained(PTEN_MODEL)
model = AutoModelForSeq2SeqLM.from_pretrained(PTEN_MODEL)
```

```python
texto_pt = "Eu gosto de comer arroz."

# words -> numbers
inputs = tokenizer(texto_pt, return_tensors="pt")

# translate
outputs = model.generate(**inputs, max_new_tokens=1024)

# numbers -> words
text_en = tokenizer.decode(outputs, skip_special_tokens=True)

display(text_en)
```
